```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('Au_nanoparticle_dataset.csv')
df.head(5)
```

```python
df.columns # should check this since pandas is case sensitive
```

Create new DaataFrame including selected columns

```python
df = df[['N_total', 'N_bulk', 'N_surface', 'R_avg' ]]
df.head(20)
```

```python
df.describe()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20,5))

for i, column in enumerate(df.columns):
    axes[i].hist(df[column], bins=30)
    axes[i].set_title(column)
    axes[i].set_xlabel("Value")
    axes[i].set_ylabel("Frequency")

plt.tight_layout()# without htis labels may overlap. this adjests spacing automatically
plt.show()
```

### Pairplot

shows the relationship between every pair of features

if diagram's points go upwards then have positive relationship. there's no pattern then no relationship between those two features

kinda same as correlation, it shows the how mush related the two features as a value. this shows is those features as relationship then how it is

Pairplots help us find:



*   Patterns
*   Relationships


*   Outliers
*   Unusual distributions

```python
sns.pairplot(df)

plt.show()
```

## KDE

estimates the probability distribution of the data

Creates smooth curve


## bivariate KDE

Shows relationship between two variables

KDE and histograms only shows one variable



```python
g = sns.PairGrid(df) #Create empty grid

g.map_diag(sns.histplot, kde=True) #diagonal plots : histogram + KDE

g.map_lower(sns.kdeplot) #Lower half of plots : bivariate KDE

# relationship repeat on lower and upper parts

plt.show()
```
