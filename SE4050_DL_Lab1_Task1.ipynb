array from an exponential distribution

Expotention distribution : generating numbers based on a mathematical probability pattern

use to represent


*   waiting time
*   arrival time

They are usually


*   Positive numbers
*   Many small adn fewer large numbers





```python
ex_array = np.random.exponential(scale=1, size=(4,4)) #Scale is avg of numbers, Size is array size(row,column)
print(ex_array)
```

Uniform Distribution


*   Every value has equal probability
*   Values are evenly spread

Normal distribution


*   Bell shaped curve
*   Most values near the mean
*   Fewer extrem values

ex: Heights of people, exam marks










```python
ex_data = np.random.exponential(scale=1, size=1000)
uni_data = np.random.rand(1000)
norm_data = np.random.normal(size=1000)
```

```python
plt.figure(figsize=(10,6))

plt.hist(ex_data, bins=100, density=True, histtype="step", label="Exponential")

plt.hist(uni_data, bins=100, density=True, histtype="step", label="Uniform")

plt.hist(norm_data, bins=100, density=True, histtype="step", label="Normal")

plt.title("Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")

plt.axis([-2.5, 2.5, 0, 1.1])
plt.legend(loc = "upper left")
plt.grid()

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)
```

**Meshgrid**

Creates a coordinate grid. Each point gets X and coordinate Y

```python
Z = X**2 + Y**2

fig = plt.figure(figsize=(10,7))

ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Plot of Z = X² + Y²")

plt.show()
```
